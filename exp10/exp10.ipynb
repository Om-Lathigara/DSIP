```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread(r"D:\MARWADI\YEAR3\SEM5\DSIP\2026-Formula1-Red-Bull-Racing-RB22-001-1080.jpg")

if image is None:
    raise FileNotFoundError("Image not found. Check the image path.")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8, 5))
plt.imshow(gray, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')
plt.show()
```

```python
weighted_kernel = np.asarray(
    [[1, 1, 1],
     [1, 2, 1],
     [1, 1, 1]], dtype=np.float32
)

weighted_kernel /= 10

weighted_mean = cv2.filter2D(image, -1, weighted_kernel)

plt.figure(figsize=(8, 5))
plt.imshow(cv2.cvtColor(weighted_mean, cv2.COLOR_BGR2RGB))
plt.title('Weighted Mean Filter')
plt.axis('off')
plt.show()
```

```python
laplacian = cv2.Laplacian(gray, cv2.CV_64F)

laplacian_abs = cv2.convertScaleAbs(laplacian)

plt.figure(figsize=(8, 5))
plt.imshow(laplacian_abs, cmap='gray')
plt.title('Laplacian Edge Detection')
plt.axis('off')
plt.show()
```

```python
roberts_x = np.asarray(
    [[1, 0],
     [0, -1]], dtype=np.float32
)

roberts_y = np.asarray(
    [[0, 1],
     [-1, 0]], dtype=np.float32
)

gx = cv2.filter2D(gray.astype(np.float32), cv2.CV_32F, roberts_x)
gy = cv2.filter2D(gray.astype(np.float32), cv2.CV_32F, roberts_y)

roberts = cv2.magnitude(gx, gy)
roberts_abs = cv2.convertScaleAbs(roberts)

plt.figure(figsize=(8, 5))
plt.imshow(roberts_abs, cmap='gray')
plt.title('Roberts Edge Detection')
plt.axis('off')
plt.show()
```
