## 1. Smoothening on noisy images (salt and pepper noise) using average, gaussian and median filters

```python
# write a python program to apply smoothing filters (average, gaussian, median) on noisy images

import cv2
import numpy as np

# Folder of the images
folder = 'D:\\MARWADI\\YEAR3\\sem5\\OPE 9\\'

# Images having salt and pepper noise
image_names = ['man.png', 'knee.png']

# Size of the filter kernel (should be an odd number)
kernel_size = 5

for name in image_names:

    # Load the image
    image = cv2.imread(folder + name)

    # Check whether image was loaded successfully
    if image is None:
        print("Image not found!")
        exit()

    # Averaging filter
    kernel = np.ones((kernel_size, kernel_size), dtype=np.float32) / (kernel_size * kernel_size)
    average_image = cv2.filter2D(image, -1, kernel)

    # Gaussian filter
    gaussian_image = cv2.GaussianBlur(image, (kernel_size, kernel_size), 0)

    # Median filter
    median_image = cv2.medianBlur(image, kernel_size)

    # Display the original and smoothed images
    cv2.imshow('Original Image - ' + name, image)
    cv2.imshow('Average Filter - ' + name, average_image)
    cv2.imshow('Gaussian Filter - ' + name, gaussian_image)
    cv2.imshow('Median Filter - ' + name, median_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 3. Sharpening of a blurred image using spatial high pass filters

```python
# write a python program to simulate sharpening using spatial high pass filters on blurred image

import cv2
import numpy as np

# Load the image
image = cv2.imread('D:\\MARWADI\\YEAR3\\sem5\\DSIP\\OPE 9\\blurred.png')

# Check whether image was loaded successfully
if image is None:
    print("Image not found!")
    exit()

# Laplacian based sharpening kernel
laplacian_kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
], dtype=np.float32)
laplacian_image = cv2.filter2D(image, -1, laplacian_kernel)

# High pass sharpening kernel
highpass_kernel = np.array([
    [-1, -1, -1],
    [-1, 9, -1],
    [-1, -1, -1]
], dtype=np.float32)
highpass_image = cv2.filter2D(image, -1, highpass_kernel)

# Unsharp masking (original + weight x (original - blurred))
blurred_image = cv2.GaussianBlur(image, (5, 5), 0)
unsharp_image = cv2.addWeighted(image, 1.5, blurred_image, -0.5, 0)

# Measure of sharpness (variance of Laplacian), higher value means more edges
def sharpness(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return cv2.Laplacian(gray, cv2.CV_64F).var()

print("Original      :", sharpness(image))
print("Laplacian     :", sharpness(laplacian_image))
print("High pass     :", sharpness(highpass_image))
print("Unsharp mask  :", sharpness(unsharp_image))

# Display images
cv2.imshow("Original Image", image)
cv2.imshow("Laplacian Sharpened Image", laplacian_image)
cv2.imshow("High Pass Sharpened Image", highpass_image)
cv2.imshow("Unsharp Masked Image", unsharp_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
