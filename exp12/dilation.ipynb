```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
kernel = np.ones((5,5), np.uint8)
```

```python
image = cv2.imread("D:\MARWADI\YEAR3\SEM5\DSIP\exp12\Screenshot 2026-10-21 124514.png")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("image")
plt.axis("off")
plt.show()
```

```python
kernel = np.ones((5,5), np.uint8)
erosion = cv2.dilate(image, kernel, iterations=1)

plt.imshow(cv2.cvtColor(erosion, cv2.COLOR_BGR2RGB))
plt.title("Erosion")
plt.axis("off")
plt.show()
```

```python
dilation = cv2.erode(image, kernel, iterations=1)

plt.imshow(cv2.cvtColor(erosion, cv2.COLOR_BGR2RGB))
plt.title("Dilation")
plt.axis("off")
plt.show()
```

```python
opening = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel)

plt.imshow(cv2.cvtColor(opening, cv2.COLOR_BGR2RGB))
plt.title("Opening")
plt.axis("off")
plt.show()
```

```python
closing = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel)

plt.imshow(cv2.cvtColor(closing, cv2.COLOR_BGR2RGB))
plt.title("Closing")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

_, binary = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

horizontal = cv2.getStructuringElement(cv2.MORPH_RECT, (40, 1))
vertical = cv2.getStructuringElement(cv2.MORPH_RECT, (1, 40))

h_lines = cv2.morphologyEx(binary, cv2.MORPH_OPEN, horizontal, iterations=3)
v_lines = cv2.morphologyEx(binary, cv2.MORPH_OPEN, vertical, iterations=3)

border = cv2.add(h_lines, v_lines)
text = cv2.subtract(binary, border)

text = cv2.bitwise_not(text)

_, text = cv2.threshold(text, 127, 255, cv2.THRESH_BINARY)

text = cv2.bitwise_not(text)

plt.imshow(text, cmap="gray")
plt.title("White Text on Black Background")
plt.axis("off")
plt.show()

text[:, :40] = 0
text[:, -40:] = 0

plt.imshow(text, cmap="gray")
plt.title("White Text on Black Background")
plt.axis("off")
plt.show()
```
